```python
from pathlib import Path

from SimulationSupport.EccentricityControl.InitialOrbitalParameters import (
    initial_orbital_parameters,
)

CHECKPOINT_DIR = Path.cwd() / "Examples"

target_params = {
    "MassRatio": 8.0,
    "DimensionlessSpinA": [3.61e-14, -4.5e-15, 0.-0.8],
    "DimensionlessSpinB": [-1.1218e-12, -3.82e-14, -0.8],
    "Eccentricity": 0.0,
}

D0, Omega0, Adot0 = initial_orbital_parameters(
    target_params,
    separation = 13.378235,
    method=  "PN",
)
print("PN:", D0, Omega0, Adot0)
```

```python
D0, Omega0, Adot0 = initial_orbital_parameters(
    target_params,
    separation = 13.378235,
    method = "GPR",
    gpr_checkpoints={
        "Omega0": str(CHECKPOINT_DIR / "gpr_model_omega.pth"),
        "Adot0": str(CHECKPOINT_DIR / "gpr_model_adot.pth"),
    },
)
print("GPR:", D0, Omega0, Adot0)
```
